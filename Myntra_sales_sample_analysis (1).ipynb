```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data=pd.read_csv("/content/myntra_dataset_ByScraping.csv")
data
```

```python
#Most Sold Products
top_sold = data.sort_values(by='number_of_ratings',ascending=False).head(10)
print(top_sold[['brand_name','pants_description','number_of_ratings']])
```

```python
labels = top_sold['brand_name'] + " - " + top_sold['pants_description']
plt.figure(figsize=(12,6))
plt.barh(labels, top_sold['number_of_ratings'])
plt.xlabel("Number of Ratings")
plt.ylabel("Products")
plt.title("Top 10 Most Demanded Products")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

corr = data[['price','MRP','discount_percent','ratings','number_of_ratings']].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
