```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# ==========================================
# DATASET 1 - CUSTOMER ORDERS
# ==========================================
order_amount = [2500,1800,3200,1500,2700,2100,3400,1900,2800,2300,3100,1700,2600,2000,3500,2400,2900,1600,3300,2200]
items_purchased = [5,3,7,2,6,4,8,3,6,4,7,2,5,4,9,5,6,2,8,4]

print("CUSTOMER ORDERS")
print("MEAN:", np.mean(order_amount))
print("MEDIAN:", np.median(order_amount))
print("MODE:", pd.Series(order_amount).mode())
print("VARIANCE:", np.var(order_amount))
print("STANDARD DEVIATION:", np.std(order_amount))
print("CORRELATION:", np.corrcoef(order_amount, items_purchased))

df = pd.DataFrame({'Order Amount': order_amount, 'Items Purchased': items_purchased})
print(df)

plt.figure(figsize=(10,5))
sns.scatterplot(x='Items Purchased', y='Order Amount', data=df)
plt.title("Customer Orders - Scatter Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(x='Items Purchased', y='Order Amount', data=df)
plt.title("Customer Orders - Line Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x='Items Purchased', y='Order Amount', data=df)
plt.title("Customer Orders - Bar Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(x='Order Amount', data=df)
plt.title("Customer Orders - Histogram")
plt.show()

plt.figure(figsize=(10,5))
plt.pie(df['Items Purchased'], labels=df['Items Purchased'], autopct='%1.1f%%')
plt.title("Customer Orders - Pie Chart")
plt.show()


# ==========================================
# DATASET 2 - EMPLOYEE SALARY
# ==========================================
salary = [35000,42000,39000,45000,37000,41000,38000,47000,36000,43000,40000,44000,39500,46000,37500,42500,38500,45500,40500,43500]
experience = [2,5,4,6,3,5,3,7,2,5,4,6,4,7,3,5,3,6,4,5]

print("EMPLOYEE SALARY")
print("MEAN:", np.mean(salary))
print("MEDIAN:", np.median(salary))
print("MODE:", pd.Series(salary).mode())
print("VARIANCE:", np.var(salary))
print("STANDARD DEVIATION:", np.std(salary))
print("CORRELATION:", np.corrcoef(salary, experience))

df = pd.DataFrame({'Salary': salary, 'Experience': experience})
print(df)

plt.figure(figsize=(10,5))
sns.scatterplot(x='Experience', y='Salary', data=df)
plt.title("Employee Salary - Scatter Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(x='Experience', y='Salary', data=df)
plt.title("Employee Salary - Line Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x='Experience', y='Salary', data=df)
plt.title("Employee Salary - Bar Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(x='Salary', data=df)
plt.title("Employee Salary - Histogram")
plt.show()

plt.figure(figsize=(10,5))
plt.pie(df['Experience'], labels=df['Experience'], autopct='%1.1f%%')
plt.title("Employee Salary - Pie Chart")
plt.show()


# ==========================================
# DATASET 3 - PRODUCT SALES
# ==========================================
units_sold = [25,80,60,30,15,40,22,35,50,75,65,18,55,48,70,12,20,28,45,16]
revenue = [1250000,80000,120000,450000,225000,160000,88000,175000,300000,75000,65000,360000,1100000,144000,210000,480000,200000,280000,450000,96000]

print("PRODUCT SALES")
print("MEAN:", np.mean(units_sold))
print("MEDIAN:", np.median(units_sold))
print("MODE:", pd.Series(units_sold).mode())
print("VARIANCE:", np.var(units_sold))
print("STANDARD DEVIATION:", np.std(units_sold))
print("CORRELATION:", np.corrcoef(units_sold, revenue))

df = pd.DataFrame({'Units Sold': units_sold, 'Revenue': revenue})
print(df)

plt.figure(figsize=(10,5))
sns.scatterplot(x='Units Sold', y='Revenue', data=df)
plt.title("Product Sales - Scatter Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(x='Units Sold', y='Revenue', data=df)
plt.title("Product Sales - Line Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x='Units Sold', y='Revenue', data=df)
plt.title("Product Sales - Bar Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(x='Units Sold', data=df)
plt.title("Product Sales - Histogram")
plt.show()

plt.figure(figsize=(10,5))
plt.pie(df['Units Sold'], labels=df['Units Sold'], autopct='%1.1f%%')
plt.title("Product Sales - Pie Chart")
plt.show()


# ==========================================
# DATASET 4 - HOSPITAL PATIENTS
# ==========================================
bill_amount = [12000,8500,15000,9200,11000,13400,7600,9800,14200,8900,10100,12500,9700,11300,13600,8800,11900,9500,14000,10800]
days_admitted = [4,2,5,3,4,5,2,3,5,2,3,4,3,4,5,2,4,3,5,3]

print("HOSPITAL PATIENTS")
print("MEAN:", np.mean(bill_amount))
print("MEDIAN:", np.median(bill_amount))
print("MODE:", pd.Series(bill_amount).mode())
print("VARIANCE:", np.var(bill_amount))
print("STANDARD DEVIATION:", np.std(bill_amount))
print("CORRELATION:", np.corrcoef(bill_amount, days_admitted))

df = pd.DataFrame({'Bill Amount': bill_amount, 'Days Admitted': days_admitted})
print(df)

plt.figure(figsize=(10,5))
sns.scatterplot(x='Days Admitted', y='Bill Amount', data=df)
plt.title("Hospital Patients - Scatter Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(x='Days Admitted', y='Bill Amount', data=df)
plt.title("Hospital Patients - Line Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x='Days Admitted', y='Bill Amount', data=df)
plt.title("Hospital Patients - Bar Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(x='Bill Amount', data=df)
plt.title("Hospital Patients - Histogram")
plt.show()

plt.figure(figsize=(10,5))
plt.pie(df['Days Admitted'], labels=df['Days Admitted'], autopct='%1.1f%%')
plt.title("Hospital Patients - Pie Chart")
plt.show()


# ==========================================
# DATASET 5 - LIBRARY RECORDS
# ==========================================
copies_issued = [12,8,15,10,9,11,7,13,6,14,10,9,8,12,11,7,6,9,5,8]
fine_amount = [120,80,150,100,90,110,70,130,60,140,100,90,80,120,110,70,60,90,50,80]

print("LIBRARY RECORDS")
print("MEAN:", np.mean(copies_issued))
print("MEDIAN:", np.median(copies_issued))
print("MODE:", pd.Series(copies_issued).mode())
print("VARIANCE:", np.var(copies_issued))
print("STANDARD DEVIATION:", np.std(copies_issued))
print("CORRELATION:", np.corrcoef(copies_issued, fine_amount))

df = pd.DataFrame({'Copies Issued': copies_issued, 'Fine Amount': fine_amount})
print(df)

plt.figure(figsize=(10,5))
sns.scatterplot(x='Copies Issued', y='Fine Amount', data=df)
plt.title("Library Records - Scatter Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(x='Copies Issued', y='Fine Amount', data=df)
plt.title("Library Records - Line Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x='Copies Issued', y='Fine Amount', data=df)
plt.title("Library Records - Bar Plot")
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(x='Copies Issued', data=df)
plt.title("Library Records - Histogram")
plt.show()

plt.figure(figsize=(10,5))
plt.pie(df['Copies Issued'], labels=df['Copies Issued'], autopct='%1.1f%%')
plt.title("Library Records - Pie Chart")
plt.show()
```
