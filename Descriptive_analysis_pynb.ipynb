```python
import statistics
import matplotlib.pyplot as plt
member_names=["Mohan","Harsh","kumar","Reddy","KumarRaja"]
age=[22,25,24,23,26]
workout_hours=[2,1.5,2.5,1,3]
calories_burned=[500,350,600,300,750]
print("------Fitess center Dataset----")

for i in range(len(member_names)):
  print(
      member_names[i],
      age[i],
      workout_hours[i],
      calories_burned[i]
      )
print("Dataset loded successfully ")
avg_workout=sum(workout_hours)/len(workout_hours)
print("\nAverage workout hours:",avg_workout)

print("Maximum calories:",max(calories_burned))
print("Minimum calories:",min(calories_burned))
print("Median calories:",statistics.median(calories_burned))
print("standard Deviation:",statistics.stdev(calories_burned))

index=calories_burned.index(max(calories_burned))
print("Most Active Member:",member_names[index])

#ploting
plt.bar(member_names,calories_burned)
plt.xlabel("Member Names")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis")
plt.show()


```
