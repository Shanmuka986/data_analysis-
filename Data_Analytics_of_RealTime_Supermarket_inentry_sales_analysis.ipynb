```python
#Super market inventory and sales analysis
import pandas as pd
import matplotlib.pyplot as plt


df = pd.DataFrame(
    {
        "Product": ["Milk", "Bread", "Rice", "Eggs", "Juice",
                    "Biscuits", "Sugar", "Butter", "Apples", "Soap"],

        "Daily_Sales": [50, 80, 40, 60, 35,
                        45, 30, 25, 55, 20],

        "Product_Stock": [200, 150, 300, 120, 100,
                          180, 250, 90, 140, 0],

        "Expiry_Date": ["2026-06-05", "2026-06-02", "2027-01-15",
                        "2026-06-10", "2026-07-20",
                        "2026-12-01", "2027-05-01",
                        "2026-08-15", "2026-06-08", "2028-01-01"],

        "Seasonal_Demand": ["High", "Medium", "Low", "High", "Medium",
                            "Low", "Medium", "High", "High", "Low"]
    }
)

print(df)


#Fast Moving Products
print("--------------------------------------------------------------------------------------")
fast_moving = df.nlargest(3, "Daily_Sales")

print("Fast Moving Products:")
print(fast_moving[["Product", "Daily_Sales"]])
print("--------------------------------------------------------------------------------------")
#  Slow Moving Products
slow_moving = df.nsmallest(3, "Daily_Sales")

print("\nSlow Moving Products:")
print(slow_moving[["Product", "Daily_Sales"]])
print("--------------------------------------------------------------------------------------")
# Out of Stock Products
out_of_stock = df[df["Product_Stock"] == 0]

print("\nOut of Stock Products:")
print(out_of_stock[["Product", "Product_Stock"]])
print("--------------------------------------------------------------------------------------")
```
