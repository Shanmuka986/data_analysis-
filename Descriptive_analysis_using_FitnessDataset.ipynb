```python


import statistics
import matplotlib.pyplot as plt
member_names=["Mohan","Harsh","kumar","Reddy","KumarRaja"]
age=[22,25,24,23,26]
workout_hours=[2,1.5,2.5,1,3]
calories_burned=[500,350,600,300,750]
print("------Fitess center Dataset----")

for i in range(len(member_names)):
  print(
      member_names[i],
      age[i],
      workout_hours[i],
      calories_burned[i]
      )
print("Dataset loded successfully ")
avg_workout=sum(workout_hours)/len(workout_hours)
print("\nAverage workout hours:",avg_workout)

print("Maximum calories:",max(calories_burned))
print("Minimum calories:",min(calories_burned))
print("Median calories:",statistics.median(calories_burned))
print("standard Deviation:",statistics.stdev(calories_burned))

index=calories_burned.index(max(calories_burned))
print("Most Active Member:",member_names[index])

#ploting
plt.bar(member_names,calories_burned)
plt.xlabel("Member Names")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


data=pd.read_csv("/content/gym_members_exercise_tracking_synthetic_data.csv")

data
```

```python
#average calories
avg_calories=data["Calories_Burned"].mean()
print("Average Calories Burned:",avg_calories)

#maximum calories
max_calories=data["Calories_Burned"].max()
print("Maximum Calories Burned:",max_calories)

#minimum calories
min_calories=data["Calories_Burned"].min()
print("Minimum Calories Burned:",min_calories)

#median calories
median_calories=data["Calories_Burned"].median()
print("Median Calories Burned:",median_calories)

#standard deviation
std_calories=data["Calories_Burned"].std()
print("Standard Deviation Calories Burned:",std_calories)

# ploting for Session_Duration (hours) and Calories_Burned
plt.scatter(data["Session_Duration (hours)"],data["Calories_Burned"])
plt.xlabel("Session Duration (hours)")
plt.ylabel("Calories Burned")
plt.title("Session Duration vs Calories Burned")
plt.show()
```
