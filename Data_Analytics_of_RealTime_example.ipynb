```python
#Swiggy/Zomato Food Delivery Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df=pd.DataFrame(
    {

    "Order_Time": [10,2,3,4,6,7,12],
    "delivery_Time": [15,3,5,6,10,2,45],
    "delivery_person": ["A","B","C","D","E","F","G"],
    "areas": ["A","B","C","D","E","F","G"],
    "Customer_Rating": [5,2,1,5,2,3,4]

    }
)
df

```

```python
#average Delivery Time
Avg_delevery_time=df["delivery_Time"].mean()
print(f"Avg_delevery_time: {Avg_delevery_time:.2f}Minutes")
print("--------------------------------------------------------------------------------------")

#Peak Order Hours
peak_order_hours=df["Order_Time"].idxmax()
print(f"Peak Order Hours: {peak_order_hours}")
print("--------------------------------------------------------------------------------------")

#Most Delayed Area
most_delayed_area=df.groupby("areas")["delivery_Time"].mean().idxmax()
print(f"Most Delayed Area: {most_delayed_area}")
print("--------------------------------------------------------------------------------------")

#Top-rated delivery time
top_rated_delivery_Partner=df.groupby("delivery_person")["Customer_Rating"].mean().idxmax()
print(f"Top-rated delivery Partner: {top_rated_delivery_Partner}")
print("--------------------------------------------------------------------------------------")

#Chart Plotting
plt.bar(df['Order_Time'],df['delivery_Time'])
plt.xlabel('Order Time')
plt.ylabel('Delivery Time')
plt.title('Order Time vs Delivery Time')
plt.show()


#performance Report
print("\n")
print("--------------------------------------------------------------------------------------")
print("Performance Report:")
print(df)


#sugesting to improve ratings and delivery time


# Suggestions to Improve Ratings and Delivery Time

print("\n")
print("--------------------------------------------------------------------------------------")
print("Suggestions to Improve Ratings and Delivery Time:")

# Areas with high delivery time
area_avg_time = df.groupby("areas")["delivery_Time"].mean()

for area, time in area_avg_time.items():
    if time > Avg_delevery_time:
        print(f"- Improve delivery routes and assign more delivery partners in {area} area.")

# Delivery partners with low ratings
partner_ratings = df.groupby("delivery_person")["Customer_Rating"].mean()

for partner, rating in partner_ratings.items():
    if rating < 4:
        print(f"- Provide additional training to delivery partner {partner} to improve customer satisfaction.")

# General Suggestions
if Avg_delevery_time > 30:
    print("- Average delivery time is high. Consider increasing delivery staff during peak hours.")

print("- Implement real-time route optimization.")
print("- Provide incentives for top-performing delivery partners.")
print("- Monitor delayed orders and identify bottlenecks.")
print("- Improve communication with customers through live order tracking.")

print("--------------------------------------------------------------------------------------")
```
