```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
marks=[70,80,39,77,67,79]
hours=[2,3,4,5,6,7,]

print("MEAN",np.mean(marks))
print("MEDIAN",np.median(marks))
print("MODE",pd.Series(marks).mode())
print("Standard deviation",np.std(marks))
print("CORRRLATION",np.corrcoef(marks,hours))

df=pd.DataFrame({'marks':marks,'hours':hours})
print(df)

plt.figure(figsize=(10,5))
sns.scatterplot(x='hours',y='marks',data=df)
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(x='hours',y='marks',data=df)
plt.show()

plt.figure(figsize=(10,5))
sns.barplot(x='hours',y='marks',data=df)
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(x='marks',data=df)
plt.show()

plt.figure(figsize=(10,5))
plt.pie(df['hours'], labels=df['hours'], autopct='%1.1f%%')
plt.title('Distribution of Hours')
plt.show()
```
